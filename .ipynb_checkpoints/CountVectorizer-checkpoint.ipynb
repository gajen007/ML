```python
from sklearn.feature_extraction.text import CountVectorizer

corpus2 = ['Bird is a Peacock Bird',
           'Peacock dances very well',
           'It eats variety of seeds',
           'Cumin seed was eaten by it once']
# bag-of-words (term frequency, binary frequency)
v2 = CountVectorizer()
print (v2.fit_transform(corpus2).todense())
print (v2.vocabulary_)
#Number of sentences in that corpus=rows of matrix
#Removes the single letter words (ex: 'a' or 'I')
#Make all words lowercase
#take unique words as columns
#give their counts
```
