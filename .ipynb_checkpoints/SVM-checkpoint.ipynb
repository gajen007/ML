```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm

def plot_svm(regressor, x1, x2):
    xx = np.linspace(1, 10)
    yy = -regressor.coef_[0][0] / regressor.coef_[0][1] * xx - regressor.intercept_ / regressor.coef_[0][1]
    plt.figure()
    plt.plot(xx, yy, label='SVM')
    plt.scatter(x1, x2)
    plt.legend()
    plt.show()

x1 = [2, 6, 3, 9, 4, 10]
x2 = [3, 9, 3, 10, 2, 13]
X = np.array([[2, 3], [6, 9], [3, 3], [9, 10], [4, 2], [10, 13]])
y = [0, 1, 0, 1, 0, 1]

# Train and plot SVM
svm_model = svm.SVC(kernel='linear', C=1.0)
svm_model.fit(X, y)
plot_svm(svm_model, x1, x2)
```
