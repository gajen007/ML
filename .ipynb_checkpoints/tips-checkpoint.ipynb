```python
import pandas as pd
dataFrame = pd.read_csv("tips.csv")
print("rows:",len(dataFrame))
print("# of columns:",len(dataFrame.columns))
print("columns:",dataFrame.columns)
print(dataFrame["size"].unique())
import matplotlib.pyplot as plt
day_counts = dataFrame["smoker"].value_counts() #replace with sex & day
plt.bar(day_counts.index, day_counts.values)
plt.xlabel("Smoker?")
plt.ylabel("Number of Records")
plt.title("Number of Records with Smokers/Non-Smokers")
plt.show()
```
