```python
from sklearn.feature_extraction import DictVectorizer
corpus1 = [{'Gender': 'Male'},
           {'Gender': 'Female'},
           {'Gender': 'Transgender'},
           {'Gender': 'Male'},
           {'Gender': 'Female'}]
v1 = DictVectorizer()
print (v1.fit_transform(corpus1).toarray())
print (v1.vocabulary_) #print the Dictionary of classification
```
