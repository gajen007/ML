```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer

corpus2 = ['Bird is a Peacock Bird',
           'Peacock dances very well',
           'It eats variety of seeds',
           'Cumin seed was eaten by it once']
v2 = CountVectorizer()
print (v2.fit_transform(corpus2).todense())
print('\n')
print (v2.vocabulary_)
print('\n')
print (TfidfVectorizer().fit_transform(corpus2).todense())
#Add squared values of each element and get the root value of that SUM
#then every element divided by that rooted-sum becomes the value in TfidfVectorizer Matrix
#Sometimes there might be slight differences with decimal numbers after zero!
```
